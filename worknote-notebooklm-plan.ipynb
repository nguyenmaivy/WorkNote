# WorkNote NotebookLM Architecture & Spec-Kitty Plan

Bản kế hoạch này mô tả kiến trúc mở rộng cho WorkNote theo hướng giống NotebookLM, đồng thời đưa ra bảng task cụ thể phù hợp để dùng `spec-kitty` quản lý.

## 1. Kiến trúc đề xuất cho WorkNote NotebookLM

### 1.1. Thực thể chính
- `Notebook` / `Project`: một bộ sưu tập học tập / nghiên cứu, chứa nhiều `Source`.
- `Source`: PDF, audio, video, web page, hình ảnh, link chat, text.
- `Note`: ghi chú AI / người dùng tạo ra liên quan đến notebook hoặc source.
- `ChatSession`: cuộc hội thoại AI có ngữ cảnh notebook.
- `ProviderConfig`: cấu hình model/AI provider.
- `SearchIndex`: embedding, vector search để tìm nội dung nhanh.

### 1.2. Kiến trúc tổng quan
```
graph TD
  Client[React Frontend] -->|REST / WebSocket| API[Backend API]
  API -->|Ngữ cảnh| NotebookService[Notebook Service]
  API -->|Upload File| StorageService[Storage / Source Management]
  API -->|Text Extraction| Extractor[Text Extraction / OCR]
  API -->|Vector Embed| EmbedService[Embedding & Search]
  API -->|AI Calls| AiService[AI Provider Layer]
  NotebookService --> SearchIndex[Vector Search / Index]
  StorageService --> DB[(Database / Storage)]
  Extractor --> DB
  EmbedService --> SearchIndex
  AiService -->|Assistant / Summary / Mindmap| DB
```

### 1.3. Mô-đun chức năng chính
- `Notebook Manager`: quản lý notebook, source, note, tags.
- `Source Uploader`: upload file / link / multi-modal.
- `Context Chat`: chat AI dựa trên nội dung source hiện tại.
- `Search & Recall`: tìm kiếm bằng text và vector search.
- `AI Pipeline`: summary, quiz generation, mindmap generation, citation.
- `Audio Lab`: TTS, STT, live translation.
- `Game + Quiz`: trò chơi học tập tích hợp quiz nội dung.
- `Profile & Storage`: lưu trạng thái người dùng, notebook history.

## 3. Feature list chưa có code (đề xuất nâng cấp)

### 3.1. Tính năng NotebookLM style
- Quản lý nhiều notebook / project.
- Quản lý nguồn tư liệu (`source`) riêng biệt.
- Chat AI theo ngữ cảnh notebook.
- Vector search / recall trong toàn bộ nội dung.
- Gợi ý trích dẫn và nguồn tham khảo khi trả lời.
- Lưu lịch sử chat và phiên notebook.
- Cho phép import từ web link, Google Drive, Dropbox.
- Export summary/mindmap/quiz sang Markdown hoặc PDF.

### 3.2. Tính năng sản phẩm nâng cao
- Multi-provider AI config (Gemini, OpenAI, local).
- Password / secret config cho backend.
- Notes editor rich text / markdown.
- Phân loại nội dung source theo chủ đề.
- Lưu trạng thái `active notebook` và `active source`.
- Tự động tạo mindmap từ nội dung text.
- Một `workspace` dashboard hiển thị tất cả notebook.

### 3.3. Tính năng UX/ổn định
- Progress bar cho upload / xử lý AI.
- Thông báo lỗi rõ ràng khi hết API key hoặc quá tải.
- Responsive UI cho desktop + mobile.
- Component `NotebookSidebar`, `SourceViewer`, `ContextChat`.
- Hỗ trợ user profile và cấu hình cá nhân.
- Lưu tạm nội dung khi mạng yếu / offline.

### 3.4. Tính năng mở rộng đồ chơi / gamification
- Quiz game dựa trên nội dung source.
- Bảng điểm / reward cho người học.
- Task checklist học tập trong notebook.
- Chế độ review tài liệu tự động.

## 4. Gợi ý làm việc với spec-kitty

### 4.1. Thêm mission / task cụ thể
- `Mission: WorkNote NotebookLM Architecture`
  - Task: Thiết kế data model notebook / source / note.
  - Task: Viết API spec cho notebook + source + chat-context.
  - Task: Định nghĩa flow xử lý AI từ upload -> extract -> chat.

### 4.2. Quy trình code / review
- Dùng `spec-kitty next` để sinh task từ spec.
- Dùng `spec-kitty implement` cho các task backend và frontend.
- Dùng `spec-kitty review` để kiểm tra task đã hoàn thành.

### 4.3. Tổ chức theo ưu tiên
- P0: refactor kiến trúc + API notebook.
- P0: UI notebook sidebar + Source Manager.
- P1: Context chat + search.
- P1: Summary / mindmap / quiz generation.
- P2: provider config, history, export.
- P2: UX polish, responsive, error handling.
